# Unit 5 · Combining models (PBL-III)
CART, bagging, random forest, AdaBoost and gradient boosting written from scratch and checked against scikit-learn; Bayesian model averaging over 64 Bayesian logistic models; a GAM; voting and stacking. Every model is scored on the same 5 folds as Unit 2.

```python
import sys, warnings
warnings.filterwarnings('ignore')
sys.path.insert(0, '..')
import pandas as pd
from IPython.display import Image, display
from earlydrop import config as C, preprocess as P
df = P.load_clean()
print(df.shape, df[C.TARGET].mean().round(3))
```

```python
from earlydrop import unit5_ensembles as U5
r = U5.run(df)   # about a minute on a laptop CPU (uses every core)
pd.DataFrame(r['compare_cv'])[['model', 'family', 'auc_fold_mean', 'auc_fold_std', 'accuracy', 'recall', 'brier', 'fit_seconds']]
```

## Scratch vs library

```python
for k in ('scratch_vs_sklearn_depth4', 'scratch_vs_sklearn_full'):
    print('CART', k, r['cart'][k])
print('AdaBoost', r['adaboost']['scratch_vs_sklearn'])
print('Gradient boosting', r['gboost']['scratch_vs_sklearn'])
print('Random forest test AUC scratch vs sklearn', r['forest']['test_auc_scratch_rf'], r['forest']['test_auc_sklearn_rf'])
```

## Is any ensemble significantly better than logistic regression?
Repeated 5 × 3 CV with the Nadeau–Bengio corrected resampled t-test.

```python
pd.DataFrame(r['significance'])
```
